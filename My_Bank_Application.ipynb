```python
import string as st
from getpass import getpass
def valid_password(password):
    if len(password) > 8:
        upper = 0
        special_charchter = 0
        digit = 0
        space = 0
        i = 0
        while i < len(password):
            if password[i].isupper():
                upper += 1

            elif password[i].isdigit():
                digit += 1
                
            elif password[i].isspace():
                space += 1

            elif password[i]in st.punctuation:
                special_charchter += 1


            i += 1
        if(upper > 0 and digit > 0 and special_charchter > 0) and (space == 0):
            return True
        else:
            return False
    else:
        False

data = {}
key = 1001

def signup():
    global key
    print("\n--- Signup ---")
    user = input("enter your username: ")
    password = getpass("Enter your password: ")
    again_password = getpass("Re Enter your Password")

    if password != again_password:
        print("password do not match")
        return

    if not valid_password(password):
        print("Password must be at least 8 chars, contain digit, uppercase, special char, and no space")
        return

    init_balance = int(input("Enter Initial Blance (>= 2000)"))
    if init_balance < 2000:
        print("minimum balance should be 2000")
        return

    data[key] = {"Name" : user , "Password" : password , "Balance" : init_balance}
    print(f"Account Created Succesfully! your Account No: {key}")
    key += 1

def login():
    print("\n--- Login ---")
    ac_no = int(input("Enter your Account Number: "))
    if ac_no in data:
        password  = getpass("Enter your password:")
        if password == data[ac_no]["Password"]:
            print("Login Successfully!")
            
            while True:
                print("\n--- Banking Services ---")
                print("1. Credit\n2. Debit\n3. Change Password\n4. Check Balance\n5. Show Details\n6. Logout")
                choice = int(input("Enter Your Choice:"))

                if choice == 1:
                    amount = int(input("Enter amount to Credit"))
                    data[ac_no]["Balance"] += amount
                    print("Amount Credit Successfully!")

                elif choice == 2:
                    amount = int(input("Enter amount to Debit"))
                    if amount <= data[ac_no]["Balance"]:
                        data[ac_no]["Balance"] -= amount
                        print("Amount Debited Successfully!")
                    else:
                        print("Insufficient balance")

                elif choice == 3:
                    new_password = getpass("Enter New Password:")
                    if valid_password(new_password):
                        data[ac_no]["Password"] = new_password
                        print("Password Change Successfully")
                    else:
                        print("Week Password Follow rules!")

                elif choice == 4:
                    print(f"Current Balance: {data[ac_no]["Balance"]}")

                elif choice == 5:
                    print(f"\n--- Account Details ---")
                    print(f"Name: {data[ac_no]['Name']}")
                    print(f"Account Number: {ac_no}")
                    print(f"Balance: {data[ac_no]['Balance']}")

                elif choice == 6:
                    print("Logout Succesfully!")
                    break
                else:
                    print("Invalid Choice")
    else:
        print("Account Does Not Exist")

def Forgot_password():
    print("\n--- Forgot Password ---")
    ac_no = int(input("Enter Your account number: "))
    if ac_no in data:
        uname = input("Enter Your Username:")
        if uname == data[ac_no]["Name"]:
            new_pass = getpass("Enter New Password: ")
            if valid_password(new_pass):
                data[ac_no]["Password"] = new_pass
                print("Password Reset Successfully")
            else:
                print("Weak password")
        else:
            print("Username mismatch!")
        
    else:
        print("Account does not exist")

def main():
    while True:
        print("\n====== Welcome to Bank Application ======")
        print("1. Login\n2. Signup\n3. Forgot Password\n4. Exit")
        ch = int(input("Enter your choice: "))
        
        if ch == 1:
            login()
        elif ch == 2:
            signup()
        elif ch == 3:
            forgot_password()
        elif ch == 4:
            print("Thank you for using Our Bank Application!")
            break
        else:
            print("Invalid choice, try again!")
main()
```
